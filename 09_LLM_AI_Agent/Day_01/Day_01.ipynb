```python
import os
from openai import OpenAI


os.environ["OPENAI_API_KEY"] = ""
```

## 1. 모델 생성

```python
client = OpenAI()

def create_email_prompt(recipient, tone, email):
  prompt = f'''
#Role
당신은 세련되고 격식있는 문체를 구사하는 **비즈니스 커뮤니케이션 전문가** 입니다.
사용자가 제공한 초안을 목적과 대상에 맞는 완벽한 비즈니스 이메일로 변환합니다.

#Context
사용자가 아래의 정보를 제공할 것입니다.
**원래 내용** : 수정이 필요한 이메일의 초안입니다.
**수신인** : 이메일을 받는 사람의 직급이나 관계입니다.
**원하는 톤** : 작성하고자 하는 이메일의 분위기입니다.

#Task
1. **문법 및 맞춤법** : 모든 오류를 완벽하게 교정합니다.
2. **비즈니스 에티켓** : 상황에 적합한 격식있는 인삿말과 맺음말을 추가합니다.
3. **톤 최적화** : 사용자가 지정한 {tone} 에 맞춰 어휘와 어투를 정밀하게 조정합니다.
4. **무결성 유지** : 원래 내용의 핵심메시지와 사실 관계는 절대 변경하지 마시오.

#Input Data
**원래 내용** : {email}
**수신인** : {recipient}
**원하는 톤** : {tone}

#Output Format
**[교정된 이메일 안]**
(여기에 최종 내용을 작성하시오)

---
**[수정 포인트]**
(어떤 부분이 비즈니스 적으로 개선되었는지 2~3개 포인트로 요약해서 설명하시오)

  '''
  return prompt
```

```python
prmpt = create_email_prompt("팀장님", "정중하게", "내일 회사 모까, ㅇ파주게써, 병원가따와서 전화하께")
```

```python
resp = client.chat.completions.create(
    model = 'gpt-4o-mini',
    messages = [{
            'role' : 'user',
            'content' : prmpt}],
    temperature=0
    )
print(resp.choices[0].message.content)
```
